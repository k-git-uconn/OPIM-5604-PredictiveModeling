```python
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Load data & add Group A (Figure 2) code
# Rule: keep only ID, drop all others
print("Load data and process Group A variables (Figure 2)")

df = pd.read_csv('listings.csv', encoding='latin1')
print(f"Original data shape: {df.shape}")

# Data Exploration / Initial QA
print("Initial data exploration")

print("\n[1] First 5 rows:")
display(df.head())

print("\n[2] Data shape:")
print(df.shape)

print("\n[3] Data types:")
display(df.dtypes.to_frame("dtype"))

print("\n[4] Missing values:")
missing_report = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(2)
})
display(missing_report[missing_report["missing_count"] > 0].sort_values("missing_count", ascending=False))

print("\n[5] Duplicate rows:")
print(df.duplicated().sum())

print("\n[6] Numeric summary:")
display(df.describe().T)
```

```python

# Data Exploration / Visual Checks
# These plots are for exploration only and do not modify the dataset.
numeric_for_plots = ['price', 'bedrooms', 'beds', 'minimum_nights',
                      'maximum_nights', 'availability_365',
                      'review_scores_location', 'calculated_host_listings_count',
                      'reviews_per_month']
numeric_for_plots = [c for c in numeric_for_plots if c in df.columns]

## Convert price to numeric values for visualization.
price_numeric = (
    df['price']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
)

price_numeric = pd.to_numeric(
    price_numeric,
    errors='coerce'
).dropna()

# Price distribution.
plt.figure(figsize=(8, 4))
plt.hist(price_numeric, bins=30)

plt.title('Price Distribution')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()


# Price boxplot.
plt.figure(figsize=(8, 2.5))
plt.boxplot(price_numeric, vert=False)

plt.title('Price Boxplot')
plt.xlabel('Price')
plt.show()

# Compact boxplots for selected variables; log scale helps variables with extreme values.
plot_cols = [c for c in ['bedrooms', 'beds', 'minimum_nights', 'maximum_nights',
                         'availability_365', 'review_scores_location'] if c in df.columns]
for col in plot_cols:
    plt.figure(figsize=(8, 2.5))
    sns.boxplot(x=pd.to_numeric(df[col], errors='coerce'))
    plt.title(f'{col} Boxplot')
    plt.xlabel(col)
    plt.show()
```

```python
# Group A columns explicitly required to be dropped in Figure 2
group_a_drop_cols = [
    'listing_url', 'scrape_id', 'last_scraped', 'source', 'name',
    'description', 'neighborhood_overview', 'picture_url',
    'host_id', 'host_url', 'host_profile_id', 'host_profile_url'
]

# Check and perform deletion
existing_a_drops = [col for col in group_a_drop_cols if col in df.columns]
df.drop(columns=existing_a_drops, inplace=True)
print(f"Dropped Group A irrelevant columns ({len(existing_a_drops)} columns), keeping only 'id'")

# Drop duplicate rows
dup_count = df.duplicated().sum()
df.drop_duplicates(inplace=True)
print(f"Duplicate rows dropped: {dup_count}")
```

```python
# Group B - Host-related variable processing
print("Processing Group B host-related variables")

# 2.1 Process host_about -> host_has_about
df['host_has_about'] = (
    df['host_about']
    .fillna('')
    .str.strip()
    .str.len() > 1
).astype(int)
df.drop(columns=['host_about'], inplace=True)

# 2.2 Process host_location
def group_host_location(location):
    if pd.isna(location): return 'Unknown'
    if 'Thessaloniki' in location: return 'Thessaloniki'
    if location.strip().endswith('Greece'): return 'Other_Greece'
    return 'Abroad'

df['host_location_group'] = df['host_location'].apply(group_host_location)
location_dummies = pd.get_dummies(df['host_location_group'], prefix='host_location', drop_first=True, dtype=int)
df = pd.concat([df, location_dummies], axis=1)
df.drop(columns=['host_location', 'host_location_group'], inplace=True)

# 2.3 Process host_is_superhost (t/f -> 1/0)
if df['host_is_superhost'].isna().any():
    df['host_is_superhost'] = df['host_is_superhost'].fillna(df['host_is_superhost'].mode()[0])
df['host_is_superhost'] = df['host_is_superhost'].map({'t': 1, 'f': 0})

# 2.4 Process time-period variables (combine into months)
df['hosts_time_as_user_total_months'] = df['hosts_time_as_user_years'] * 12 + df['hosts_time_as_user_months']
df['hosts_time_as_host_total_months'] = df['hosts_time_as_host_years'] * 12 + df['hosts_time_as_host_months']
df.drop(columns=['hosts_time_as_user_years', 'hosts_time_as_user_months',
                 'hosts_time_as_host_years', 'hosts_time_as_host_months'], inplace=True)
```

```python

# Group C - Missing values and basic cleaning
print("Processing Group C missing values")

# Drop columns with missing rate > 80% (but do not drop columns required by Figure 1, such as host_since, amenities, etc.)
missing_ratio = df.isnull().mean()
high_missing_cols = missing_ratio[missing_ratio > 0.8].index.tolist()
# Protect columns required to be kept in Figure 1
protected_cols = ['host_since', 'amenities', 'availability_365']
high_missing_cols = [col for col in high_missing_cols if col not in protected_cols]

df.drop(columns=high_missing_cols, inplace=True)
print(f"Columns with >80% missing values dropped: {high_missing_cols}")

# Fill numeric missing values (median)
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values (mode)
cat_cols = df.select_dtypes(include=['object', 'category']).columns
for col in cat_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Step 4: Group D - Attributes, amenities, and price cleaning
print("Step 4: Processing Group D attributes and price")

# 4.1 bathrooms
# Check if 'bathrooms_text' still exists before processing (it might have been processed by another cell)
if 'bathrooms_text' in df.columns:
    extracted = df['bathrooms_text'].str.extract(r'([\d.]+)')[0]
    extracted = pd.to_numeric(extracted, errors='coerce').fillna(0.5)
    df['bathrooms'] = df['bathrooms'].fillna(extracted)
    df.drop(columns=['bathrooms_text'], inplace=True)

# 4.2 bedrooms/beds
# Ensure 'bedrooms' and 'beds' are numeric before applying median imputation
df['bedrooms'] = pd.to_numeric(df['bedrooms'], errors='coerce')
df['beds'] = pd.to_numeric(df['beds'], errors='coerce')
df['bedrooms'] = df['bedrooms'].fillna(df.groupby('accommodates')['bedrooms'].transform('median'))
df['beds'] = df['beds'].fillna(df.groupby('accommodates')['beds'].transform('median'))

# 4.3 amenities: Figure 1 requires keeping the amenities column
# To keep the original amenities column, do not drop it here; if a count feature is also needed, keep amenities_count separately
df['amenities_count'] = df['amenities'].str.count(',') + 1
df['amenities_count'] = pd.to_numeric(df['amenities_count'], errors='coerce')

# 4.4 price (Target)
df['price'] = df['price'].replace(r'[\$,]', '', regex=True)
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df.dropna(subset=['price'], inplace=True)

# Remove extreme outliers
df = df[df['price'] < 5000]

# 4.5 Drop redundant price-related columns
price_related_drops = ['price_quote_checkin_date', 'price_quote_checkout_date',
                       'price_quote_total_price', 'price_quote_price_per_night', 'price_quote_raw']
df.drop(columns=[c for c in price_related_drops if c in df.columns], inplace=True)
```

```python
import pandas as pd
# Step 5: Group E - Duration and availability cleaning
print("\n" + "="*60)
print("Step 5: Processing Group E availability variables")
print("="*60)

# Keep minimum_nights, maximum_nights, has_availability, availability_30/60/90/365
# Handle outliers (upper-bound clipping)
for col in ['minimum_nights', 'maximum_nights']:
    # Ensure the column is numeric, coercing errors to NaN
    df[col] = pd.to_numeric(df[col], errors='coerce')
    # Fill any NaN values (including those from coercion) with the column's median
    if df[col].isna().any():
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    upper_bound = q3 + 1.5 * iqr
    df[col] = df[col].clip(upper=upper_bound)

# Convert has_availability to 0/1
if df['has_availability'].isna().any():
    df['has_availability'] = df['has_availability'].fillna(df['has_availability'].mode()[0])
df['has_availability'] = df['has_availability'].map({'t': 1, 'f': 0})

# Drop redundant time-period columns
e_drop_cols = ['minimum_minimum_nights', 'maximum_minimum_nights',
               'minimum_maximum_nights', 'maximum_maximum_nights',
               'minimum_nights_avg_ntm', 'maximum_nights_avg_ntm']
df.drop(columns=[c for c in e_drop_cols if c in df.columns], inplace=True)
```

```python
# Step 6: Group G - Ratings and host scale
print("Step 6: Processing Group G ratings and scale")

# Fill missing review_scores values
review_cols = ['review_scores_location', 'review_scores_cleanliness',
               'review_scores_checkin', 'review_scores_communication', 'review_scores_value']

# Convert 'number_of_reviews' to numeric, coercing errors, and fill NaN with 0
df['number_of_reviews'] = pd.to_numeric(df['number_of_reviews'], errors='coerce').fillna(0)
df['has_reviews'] = (df['number_of_reviews'] > 0).astype(int)

for c in review_cols:
    if c in df.columns:
        # Convert review score columns to numeric, coercing errors
        df[c] = pd.to_numeric(df[c], errors='coerce')
        df[c] = df[c].fillna(df[c].median())

# Fill missing reviews_per_month values
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
```

```python
# Filter and order final columns according to Figure 1 (Ref. Sheet 04 > Column G)

print("Arranging final column order according to Figure 1")

# Define the list strictly according to the order in Figure 1
target_columns_fig1 = [
    'id',
    'host_name',
    'host_since',
    'property_type',
    'room_type',
    'accommodates',
    'bathrooms',
    'bedrooms',
    'beds',
    'amenities',
    'price', # Target
    'minimum_nights',
    'maximum_nights',
    'availability_365',
    'review_scores_location',
    'calculated_host_listings_count',
    'reviews_per_month'
]

# Check availability and warn
available_cols = [col for col in target_columns_fig1 if col in df.columns]
missing_cols = [col for col in target_columns_fig1 if col not in df.columns]

if missing_cols:
    print(f"Warning: The following columns required by Figure 1 are not in the current data. Please check the previous processing logic: {missing_cols}")

# Keep only the columns required by Figure 1, strictly preserving the order
df_final = df[available_cols].copy()

print(f"\nFiltering complete! The final data contains {len(df_final.columns)} columns, in the exact order of Figure 1.")
print(df_final.columns.tolist())
```

```python

# Final Data Quality Audit
# IMPORTANT: This section checks the data but does NOT change df_final.
# This preserves the existing transformation logic and makes the notebook
# easier to document in the paper.
print("Final data quality audit")

print("\n[1] Final shape:")
print(df_final.shape)

print("\n[2] First 5 rows:")
display(df_final.head())

print("\n[3] Final dtypes:")
display(df_final.dtypes.to_frame("dtype"))

print("\n[4] Missing values:")
final_missing = pd.DataFrame({
    "missing_count": df_final.isna().sum(),
    "missing_pct": (df_final.isna().mean() * 100).round(2)
})
display(final_missing[final_missing["missing_count"] > 0].sort_values("missing_count", ascending=False))

print("\n[5] Duplicate rows:")
print(df_final.duplicated().sum())

print("\n[6] Numeric summary:")
display(df_final.describe().T)

# IQR-based outlier report. This is diagnostic only; it does not alter df_final.
def iqr_outlier_report(dataframe):
    rows = []
    numeric_cols = dataframe.select_dtypes(include=np.number).columns
    for col in numeric_cols:
        s = dataframe[col].dropna()
        if s.empty:
            continue
        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        count = ((s < lower) | (s > upper)).sum()
        rows.append([col, q1, q3, lower, upper, int(count)])
    return pd.DataFrame(
        rows,
        columns=["column", "Q1", "Q3", "lower_bound", "upper_bound", "outlier_count"]
    ).sort_values("outlier_count", ascending=False)

print("\n[7] IQR outlier check (diagnostic only):")
display(iqr_outlier_report(df_final))

print("\n[8] Categorical columns / number of unique values:")
cat_summary = pd.DataFrame({
    "dtype": df_final.dtypes.astype(str),
    "unique_values": df_final.nunique(dropna=False)
})
display(cat_summary)
```

```python
# Final Cleaning: Missing Values and Outliers
# 1. Handle missing values in numeric columns.
numeric_cols = df_final.select_dtypes(include='number').columns

for col in numeric_cols:
    if df_final[col].isnull().any():
        df_final[col] = df_final[col].fillna(
            df_final[col].median()
        )


# 2. Handle missing values in categorical columns.
categorical_cols = df_final.select_dtypes(
    include=['object', 'category']
).columns

for col in categorical_cols:
    if df_final[col].isnull().any():
        df_final[col] = df_final[col].fillna(
            df_final[col].mode()[0]
        )

# Drop host_since because all values are missing.
df_final.drop(columns=['host_since'], inplace=True, errors='ignore')

# Ensure 'availability_365' is numeric before outlier handling
if 'availability_365' in df_final.columns:
    df_final['availability_365'] = pd.to_numeric(df_final['availability_365'], errors='coerce')
    # Fill any NaN values that might result from coercion or were initially missing
    if df_final['availability_365'].isnull().any():
        df_final['availability_365'] = df_final['availability_365'].fillna(df_final['availability_365'].median())

# Ensure 'reviews_per_month' is numeric before outlier handling
if 'reviews_per_month' in df_final.columns:
    df_final['reviews_per_month'] = pd.to_numeric(df_final['reviews_per_month'], errors='coerce')
    # Fill any NaN values that might result from coercion or were initially missing
    if df_final['reviews_per_month'].isnull().any():
        df_final['reviews_per_month'] = df_final['reviews_per_month'].fillna(df_final['reviews_per_month'].median())

# 3. Handle outliers in continuous numeric variables using the IQR rule.
outlier_cols = [
    'price',
    'minimum_nights',
    'maximum_nights',
    'availability_365',
    'review_scores_location',
    'reviews_per_month'
]

for col in outlier_cols:
    if col in df_final.columns:

        q1 = df_final[col].quantile(0.25)
        q3 = df_final[col].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        df_final[col] = df_final[col].clip(
            lower=lower_bound,
            upper=upper_bound
        )


# 4. Final check for missing values.
print("Total missing values:")
print(df_final.isnull().sum().sum())


# 5. Final check for outliers.
print("\nFinal outlier check:")

for col in outlier_cols:
    if col in df_final.columns:

        q1 = df_final[col].quantile(0.25)
        q3 = df_final[col].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outliers = (
            (df_final[col] < lower_bound) |
            (df_final[col] > upper_bound)
        ).sum()

        print(f"{col}: {outliers}")


# 6. Display the final shape.
print("\nFinal shape:")
print(df_final.shape)
```
